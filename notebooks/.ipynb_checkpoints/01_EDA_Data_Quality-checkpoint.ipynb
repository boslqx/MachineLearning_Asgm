```python
# EDA and data quality investigation
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/raw/diabetic_data.csv")

print("Shape:", df.shape)
display(df.head())
display(df.info())
```

```python
# inspect the dataset
df.info()
```

```python
# sort out all the features and its type
df.describe(include="all").T
```

```python
# calculate the number of unique value
df.nunique().sort_values()
```

```python
# investigate missing values
missing_count = (df == "?").sum()
missing_pct = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_pct
})

missing_summary = missing_summary[
    missing_summary["Missing_Count"] > 0
].sort_values("Missing_Percentage", ascending=False)

display(missing_summary)

plt.figure(figsize=(10, 6))

missing_summary["Missing_Percentage"].plot(kind="bar")

plt.title("Missing Values by Feature")
plt.xlabel("Feature")
plt.ylabel("Missing Percentage (%)")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# investigate duplicate value
duplicate_rows = df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)
print("Duplicate percentage:", duplicate_rows / len(df) * 100)
```

```python
# investiage identifier
print("Unique encounter IDs:", df["encounter_id"].nunique())
print("Unique patient numbers:", df["patient_nbr"].nunique())
print("Total rows:", len(df))
```

```python
# counting and summarizing visits
patient_counts = df["patient_nbr"].value_counts()

display(patient_counts.describe())
display(patient_counts.head(20))

patient_counts = df["patient_nbr"].value_counts()

display(patient_counts.describe())
display(patient_counts.head(20))
```

```python
# analyze the target
target_counts = df["readmitted"].value_counts()

target_percentage = (
    df["readmitted"]
    .value_counts(normalize=True)
    .mul(100)
)

target_summary = pd.DataFrame({
    "Count": target_counts,
    "Percentage": target_percentage
})

display(target_summary)

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="readmitted")

plt.title("Distribution of Readmission Status")
plt.xlabel("Readmission Status")
plt.ylabel("Number of Encounters")
plt.tight_layout()
plt.show()
```

```python
# create proposed binary target
df["readmitted_binary"] = (
    df["readmitted"] == "<30"
).astype(int)

binary_counts = df["readmitted_binary"].value_counts()

binary_percentage = (
    df["readmitted_binary"]
    .value_counts(normalize=True)
    .mul(100)
)

binary_summary = pd.DataFrame({
    "Count": binary_counts,
    "Percentage": binary_percentage
})

display(binary_summary)

plt.figure(figsize=(6, 5))

sns.countplot(data=df, x="readmitted_binary")

plt.title("Binary 30-Day Readmission Distribution")
plt.xlabel("Readmitted Within 30 Days (0 = No, 1 = Yes)")
plt.ylabel("Number of Encounters")
plt.tight_layout()
plt.show()
```

```python
# Investigate constant and near-constant features
unique_counts = df.nunique().sort_values()

display(unique_counts)

constant_features = [
    col for col in df.columns
    if df[col].nunique() <= 1
]

print("Constant features:")
print(constant_features)
```

```python
# investigate numerical feature
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

print(numeric_cols)
display(df[numeric_cols].describe().T)

for col in numeric_cols:
    plt.figure(figsize=(7, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.tight_layout()
    plt.show()
```

```python
# investigate outlier
for col in numeric_cols:
    plt.figure(figsize=(7, 3))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.tight_layout()
    plt.show()
```

```python
# investigate categorical variable
categorical_cols = df.select_dtypes(include="object").columns.tolist()

for col in categorical_cols:
    print(f"\n===== {col} =====")
    display(df[col].value_counts(dropna=False).head(10))
```

```python
# investigate two id column
# encounter id
print(df["encounter_id"].is_unique)

# patient nbr
print("Unique patients:", df["patient_nbr"].nunique())
print("Total encounters:", len(df))

patient_encounters = (
    df.groupby("patient_nbr")
      .size()
      .sort_values(ascending=False)
)

display(patient_encounters.head(20))
```

```python
# check duplicate row
print("Duplicate rows:", df.duplicated().sum())

print("Unique encounter IDs:", df["encounter_id"].nunique())
print("Unique patient numbers:", df["patient_nbr"].nunique())
print("Total rows:", len(df))

patient_counts = df["patient_nbr"].value_counts()

print("Patients with multiple encounters:",
      (patient_counts > 1).sum())

print("Patients with 3+ encounters:",
      (patient_counts >= 3).sum())
```

```python
display(binary_summary)
```

```python
# determine whether patients with multiple encounters can have different readmission outcomes.
patient_target_variation = (
    df.groupby("patient_nbr")["readmitted_binary"]
      .nunique()
)

print(
    "Patients with both readmission classes:",
    (patient_target_variation > 1).sum()
)

print(
    "Percentage of patients with both classes:",
    (patient_target_variation > 1).mean() * 100
)
```
